```python
my_tasks = []


while True :
    print("---------------------------")
    add = int(input("add item press 1:  \n"
                "delete item press 2: \n"
                "to see to do list press 3: \n"
                "To exit press 4: \n"))
    
    if (add == 1):
        print("---------------------------")
        while True:
            
            task = str(input("Enter you task "))
            if (task =='exit'):
                break
            my_tasks.append(task)
        print("")
        print ("Your Todo List")
        for i in range (len(my_tasks)):
            print (i+1, ".",my_tasks[i])
            
    elif (add == 2):
        print("Current list")
        for i in range (len(my_tasks)):
            print (i+1, "." , my_tasks[i])
        print("")
        print("---------------------------")
        f = int(input ("enter the number of the item you want to delete: "))
        my_tasks.remove(my_tasks[f-1])

        print("")
        print ("New Task List")
        for i in range (len(my_tasks)):
            print (i+1 , ".",my_tasks[i])

    elif(add  == 3):
        for i in range (len(my_tasks)):
            print(i+1,".",my_tasks[i])
    elif(add == 4):
        print ("bye")
        break
```
